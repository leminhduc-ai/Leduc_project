```python
n = int(input())
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)
print(fib(n))
```

```python
#W5A1
def Max_Of_Two(a:int, b:int) -> int:
    return max(a,b)

try:
    a,b = map(int, input().split())
    print(Max_Of_Two(a, b))
except ValueError:
    print("Nhập lỗi")
```

```python
#W5A2
def swap(a,b):
    a,b = b,a
    return a,b

print(*swap(12,10))
```

```python
#W5A3
import math
def prime_number(n: int) -> bool:
    for i in range(2, math.isqrt(n)+1):
        if n % i == 0:
            return False
    return (n >= 2)

n = int(input())
for x in range(2,n+1):
    if(prime_number(x)):
        print(x, end = ' ')
```

```python
#W5A4
def so_hoan_hao(n: int) -> bool:
    n = int(input())
    if n <= 1:
        return False
    tong = 1
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            tong += i
            if i != n // i:
                tong += n // i
    return tong == n
print(so_hoan_hao(n))
```

```python
#W5A5
def tra_vi_tri(lst: list[int], k: int) -> bool:
    k = int(input())
    for i in range(len(lst)):
        if lst[i] == k:
            return i+1
    return -1
lst = [1,2,3,4,5,9,5,2,4]
print(tra_vi_tri(lst, 1234546789))
```

```python
#W5A6
def giai_thua(n: int) -> bool:
    n = int(input())
    ket_qua = 1
    for i in range(1, n+1):
        ket_qua *= i
    return ket_qua
print(giai_thua(n))
```

```python
#W5A9
def tong_chu_so(n: int) -> bool:
    n = int(input())
    tong = 0
    n = abs(n)
    while n > 0:
        chu_so_cuoi = n % 10
        tong += chu_so_cuoi
        n //= 10
    return tong
print(tong_chu_so(n))
```
