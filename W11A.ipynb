```python
#W11A1
def reverse(path: str) -> int:
    with open(path, "r", encoding="utf-8") as f: 
        n_str = f.read().strip()
    reversed_str = n_str[::-1]
    return int(reversed_str)
with open("input.txt", "w", encoding="utf-8") as f: 
    f.write(str(input()))
print(reverse("input.txt"))
```

```python
#W11A2
def canWinNim (path: str) -> bool:
    with open(path, "r", encoding="utf-8") as f: 
        n = f.read()
        for i in range(4,int(n)+1,3):
            if int(n) == int(i):
                return False
        return True
with open("input.txt", "w", encoding="utf-8") as f: 
    f.write(str(input()))
print(canWinNim("input.txt"))   
```

```python
#W11A4
import string
from collections import Counter
def areAnagrams(path1: str, path2: str) -> bool:
    def normalize(text_str) -> str:
        text = text_str.lower()
        translator = str.maketrans('', '', string.punctuation + string.whitespace)
        return text.translate(translator)
    with open(path1, "r", encoding="utf-8") as f1: 
        s1 = f1.read().strip()
    with open(path2, "r", encoding="utf-8") as f2: 
        s2 = f2.read().strip()
    s1 = normalize(s1)
    s2 = normalize(s2)
    return Counter(s1) == Counter(s2)
print(areAnagrams("file1.txt", "file2.txt"))
```

```python
#W11A5
def count(path: str, character: str):
    with open(path, "r", encoding="utf-8") as f:
        sent = f.read().strip()
        return sent.count(character)

text = input()
with open("count.txt", "w", encoding="utf-8") as f:
    f.write(text)

char = input()

print(count("count.txt", char))
```

```python
#W11A9
import csv

def grade10(path: str):
    with open(path, mode='r', encoding='utf-8') as file:
        reader = csv.reader(file)
        next(reader)  

        for row in reader:
            thcs4_score = int(row[6])
            if thcs4_score == 10:
                full_name = f"{row[1]} {row[2]} {row[3]}"
                print(full_name)

sample_data = [
    ["STT","Ho","Ten dem","Ten","Ngay sinh","Dai so","THCS4","THCS1"],
    ["1","Truong","Vo","Ky","10/10/1999","8","10","9"],
    ["2","Chu","Chi","Nhuoc","13/9/1999","10","10","10"]
]

with open("input.csv", mode="w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(sample_data)

grade10("input.csv")
```

```python
#W11A15
import csv

def findMovies(path: str):
    with open(path, mode='r', encoding='utf-8') as file:
        reader = csv.reader(file)
        next(reader)  
        flag = True
        for row in reader:
            imdb_score = float(row[3])
            mo_ta = row[2]
            if imdb_score >= 8.0 and mo_ta != 'boring':
                movie = f"{row[1]}"
                print(movie)
                flag = False
        if flag: 
            print("All are boring!")
            
            
sample_data = [
    ["id","movie","description","rating"],
    ["1","Thor","great 3D","8.0"],
    ["2","The Rings","Fantasy",'7.5'],
    ['3','Irish','boring','6.2'],
    ['4','Ice song','Fantasy','8.6'],
    ['5','Home Alone','Interesting','6.5']
]

with open("input.csv", mode="w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(sample_data)

findMovies("input.csv")
```
