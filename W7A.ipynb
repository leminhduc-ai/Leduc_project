```python
#A4
arr = list(map(int, input().split()))
def mode(arr):
    cnt = {}
    for i in arr:
        cnt[i] = cnt.get(i, 0)+1
    max_cnt = max(cnt.values())
    for i in arr:
        if cnt[i] == max_cnt:
            return i, max_cnt
val, freq = mode(arr)
print(f'{val} xuat hien nhieu nhat, som nhat, {freq} lan')
```

```python
#A5
arr = list(map(int, input().split()))
X = int(input())
cnt = 0
for a in range(0,len(arr)):
    for b in range(a+1,len(arr)):
        if arr[a] + arr[b] == X:
            cnt += 1
print(cnt)
```

```python
#W7A11
arr = tuple(input().split())
w = input()
cnt = 0
for i in range(0, len(arr)):
    if w == arr[i]:
        cnt += 1
print(cnt)
```

```python
#W7A14
n = int(input())
arr = list(map(int, input().split()))
if n != len(arr):
    print("Error")
else:
    flag = False
    for i in range(0, len(arr)):
        if arr[i] == 7:
            print(i, end = ' ')
            flag = True
    if not flag:
        print("Not found")
```
