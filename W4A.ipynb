```python
#W4A13
while True:
    try:
        a,b = map(int, input().split())
        if a <= 0 or b <= 0:
            print("Nhập số nguyên dương")
        else:
            break
    except ValueError:
        print("Nhập sai")

suma = 0
for i in range(1,a):
    if a % i == 0:
        suma += i

sumb = 0
for j in range(1,b):
    if b % j == 0:
        sumb += j

if suma == b and sumb == a:
    print("true")
else:
    print("false")
```

```python
#W4A14
while True:
    try:
        m,n = map(int, input().split())
        if m <= 0 or n <= 0:
            print("Nhập số nguyên dương")
        else:
            break
    except ValueError:
        print("Nhập sai")
while n != 0:
        r = m % n
        m = n
        n = r
print(m)
```
