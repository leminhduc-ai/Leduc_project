```python
#W6A2
def tinh_tong_tich_luy(a):
    tong_tich_luy = []
    tong = 0
    for i in a:
        tong += i 
        tong_tich_luy.append(tong)
    return tong_tich_luy
a = list(map(int, input().split()))
tong_tich_luy = tinh_tong_tich_luy(a)
print(tong_tich_luy)
```

```python
#A8
def counts(a):
    cnt = {}
    for i in a:
        if i in cnt:
            cnt[i] += 1
        else:
            cnt[i] = 1
    return cnt
a = tuple(input().split())
print(counts(a))
```

```python
#A9
dict1 = eval(input())
dict2 = eval(input())
gop = dict1.copy()
for k, v in dict2.items():
    if k in gop:
        gop[k] += v
    else:
        gop[k] = v
sort_dict = dict(sorted(gop.items()))
print(sort_dict)
```

```python
#A10
lst = list(map(int, input().split()))
k = int(input())
cap = []
for i in range(len(lst)):
    for j in range(i+1, len(lst)):
        if lst[i] + lst[j] == k:
            cap.append((lst[i], lst[j]))
cap.sort(key=lambda x: x[0])
print(cap)
```
