```python
#W7A16
from collections import Counter
def min_deletions_to_beautiful(arr):
    freq = Counter(arr)
    deletions = 0

    for x in freq:
        if freq[x] == x:
            continue  
        elif freq[x] > x:
            deletions += freq[x] - x  
        else:
            deletions += freq[x]      

    return deletions
arr = list(map(int, input("Nhập mảng số nguyên không âm: ").split()))
deletions = min_deletions_to_beautiful(arr)
if deletions == 0:
    print("YES")
else:
    print("NO", deletions)
```

```python
#W7A18
arr = list(map(int, input().split()))
chan = sorted(i for i in arr if i % 2 == 0)
le = sorted([i for i in arr if i % 2 != 0], reverse=True)
ket_qua = chan + le
print(*ket_qua)
```

```python
#W7A21
N, T = map(int, input().split())
max_day = [0, 31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
if (T < 1 or T > 12) or N > max_day[T]: 
    print("invalid")
else:
    total = N
    for i in range(T):
        if i < T:
            total += max_day[i]
    print(total)
```
